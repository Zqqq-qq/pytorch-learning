利用方法一保存网络模型结构及其参数，观察左侧文件工作区的变化，加载模型并打印查看，分别使用终端命令和单元格代码查看模型大小

```python
import torch
import torchvision
import os

# 1. 实例化一个未经过预训练的 VGG16 骨架模型，现代化写法：使用官方推荐的 weights=None 替代已过时的 pretrained=False
vgg16 = torchvision.models.vgg16(weights=None)

# 2. 运行保存方式 1：同时保存网络模型结构和权重参数，后缀名通常为 .pth 或 .pt
torch.save(vgg16, 'vgg16.pth')

# 3. 运行加载模型
'''由于 PyTorch 2.6+ 将默认参数设为了 weights_only=True，直接加载会 UnpicklingError，因为本文件由我们本地创建，可信，故显式指定 weights_only=False 绕过审查并成功加载'''
model = torch.load('vgg16.pth', weights_only=False)

# 4. 打印加载后的模型结构
print(model)

# 5. 计算并打印文件大小 (单位: MB)
file_size_mb = os.path.getsize('vgg16.pth') / (1024 * 1024)
print(f"模型文件 'vgg16.pth' 的大小为: {file_size_mb:.2f} MB") # 预期输出: 527.81 MB (精确到两位小数)
```

使用`.state_dict()`保存网络模型结构及其参数，观察左侧文件工作区的变化，加载模型并打印查看，分别使用终端命令和单元格代码查看这种保存模式下的模型大小，并与方法一进行对比

```python
import torch
import torchvision

vgg16 = torchvision.models.vgg16(weights=None)

#保存方式2:保存模型参数，更轻量，官方推荐
torch.save(vgg16.state_dict(), 'vgg16_2.pth')

#加载模型
model = torch.load('vgg16_2.pth')
print(model)

import os

# 计算并打印文件大小 (单位: MB)
file_size_mb = os.path.getsize('vgg16_2.pth') / (1024 * 1024)
print(f"模型文件 'vgg16_2.pth' 的大小为: {file_size_mb:.2f} MB")
```

将第二种方式保存的模型参数恢复为完整模型结构并输出

```python
import torch
import torchvision
import os

# 1. 实例化 VGG16 未经训练的模型结构
vgg16 = torchvision.models.vgg16(weights=None)

# 2. 运行保存方式 2：只保存模型参数（state_dict），更轻量，官方推荐
torch.save(vgg16.state_dict(), 'vgg16_2.pth')

# 3. 核心：将第二种方式保存的参数，恢复加载为完整的可运行网络模型
# 第一步：实例化一个结构一模一样的空白模型
vgg16_restore = torchvision.models.vgg16(weights=None)

# 第二步：加入权重参数字典
vgg16_restore.load_state_dict(torch.load('vgg16_2.pth'))

# 4. 打印恢复后的完整模型
print(vgg16_restore)
```

自定义一个模型结构并尝试用第一种方法保存

```python
import torch
from torch import nn
from torch.nn import Conv2d, MaxPool2d, Flatten, Linear

class model(nn.Module):
    def __init__(self):
        super(model, self).__init__()
        self.model1 = nn.Sequential(
            Conv2d(in_channels=3, out_channels=32, kernel_size=5, stride=1, padding=2),
            MaxPool2d(2),
            Conv2d(in_channels=32, out_channels=32,kernel_size=5, stride=1, padding=2),
            MaxPool2d(2),
            Conv2d(in_channels=32, out_channels=64, kernel_size=5, stride=1, padding=2),
            MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(1024, 64),
            nn.Linear(64, 10)
        )

    def forward(self, x):
        x = self.model1(x)
        return x

test = model()
torch.save(test, 'test.pth')
```

观察左侧文件工作区可以看到，方法一仍能正常保存我们的自定义模型，现在尝试用`load`加载它

```python
model = torch.load('test.pth', weights_only=False)
print(model)
```
