```python
#PyTorch 极简模型代码
import torch
from torch import nn

# 1. 定义我们自己的网络类，必须继承 nn.Module（所有神经网络的基类）
class Tudui(nn.Module):

    # 构造函数：模型实例化（创建对象）时自动运行，用来声明和购买“网络层（积木零件）”
    def __init__(self):
        # 一键调用父类 nn.Module 的初始化构造函数！
        super().__init__()
        # (在这个极简模型骨架中，因为不需要任何层计算，所以这里保持空白)

    # 前向传播：定义数据的“加工流水线”，input 是传入的输入数据
    def forward(self, input):
        # 计算逻辑：让输入的张量加一。真实模型里这里会换成具体的矩阵和层计算
        output = input + 1
        # 返回计算完毕的输出数据
        return output

# 2. 运行与验证
# 1. 实例化我们的网络对象
tudui = Tudui()

# 2. 创建一个 PyTorch 浮点张量（Tensor），初始值为 1.0
x = torch.tensor(1.0)

# 3. 核心：直接像调用函数一样调用模型 tudui(x)
# 这会自动在后台触发运行类内的 forward(self, x) 方法
# 严禁写成 tudui.forward(x)，因为直接调用会错过底层梯度流和钩子（Hooks）的处理
output = tudui(x)

# 4. 打印输出最终计算结果
print(output)  # 预期输出: tensor(2.)
```
